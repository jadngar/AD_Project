# Carga del Dataset 

```python
import pandas as pd 
import numpy as np

df = pd.read_csv('sales_record.csv')
df.head()
```

# Inspección Inicial

```python
df.info()
df.describe()
df.isnull().sum()
```

# Limpieza de datos

```python
# Convertir las columnas de fecha
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], errors='coerce')

# Verificación de la conversión
df.dtypes
```

# Guardado del Dataset Limpio

```python
# Guardar el dataframe limpio en un nuevo archivo
df.to_csv('sales_record_cleaned.csv', index=False)
```

# Carga el nuevo dataset que se limpió

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Carga del dataset limpio
df = pd.read_csv('sales_record_cleaned.csv')
```

# Visualizamos los datos

```python
# Configuración del estilo de gráficos
sns.set(style="whitegrid")

# Creación de un gráfico de barras para visualizar el total de ganancias por región
plt.figure(figsize=(10, 6))
sns.barplot(x='Total Profit', y='Region', data=df, estimator=sum, errorbar=None, color='lightblue')
plt.title('Total de Ganancias por Región')
plt.xlabel('Total de Ganancias')
plt.ylabel('Región')
plt.tight_layout()
```
