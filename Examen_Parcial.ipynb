```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Cargando los datos

data_path = 'Data.csv'
data = pd.read_csv(data_path)
data
```

```python
# Gráfica de la distribución de edad y salario por País

plt.figure(figsize=(10, 6))
sns.scatterplot(data=data, x='Age', y='Salary', hue='Country', palette='Set1', s=100)
plt.title('Distribución de Edad y Salario por País')
plt.xlabel('Edad')
plt.ylabel('Salario')
plt.legend(title='País')
plt.grid(True)
plt.show()
```
